```python
#Modular Calculator Mini-Program

def add(a, b):
    return a + b


def subtract(a, b):
    return a - b


def multiply(a, b):
    return a * b


def divide(a, b):
    if b == 0:
        return "Error: Cannot divide by zero"
    return a / b


def main():
    print("Select operation:")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")

    operations = {
        "1": ("Addition", add),
        "2": ("Subtraction", subtract),
        "3": ("Multiplication", multiply),
        "4": ("Division", divide),
    }

    choice = input("Enter choice (1, 2, 3, or 4): ")

    if choice not in operations:
        print("Invalid Input! Please run the program again and select 1, 2, 3, or 4.")
        return

    try:
        a = float(input("Please enter your first desired number: "))
        b = float(input("Please enter your second desired number: "))
    except ValueError:
        print("Error: Please enter valid numbers.")
        return

    name, func = operations[choice]
    result = func(a, b)
    print("Result:", result)


if __name__ == "__main__":
    main()
```
